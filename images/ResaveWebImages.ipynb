```python
import os
from PIL import Image

# --- CONFIGURATION ---
SOURCE_FOLDER = r"E:/Projekte/2D Projekte\Website/images"  # Replace with your folder path
QUALITY = 80  # Compression rate: 1 (worst/smallest) to 100 (best/largest). 75-80 is usually the sweet spot.
DELETE_ORIGINAL = False  # Set to True if you want to delete the old images after conversion
# ---------------------

def convert_to_webp(folder_path, quality_setting, delete_old=False):
    # Supported source formats
    valid_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.tiff')
    
    # Ensure the folder exists
    if not os.path.exists(folder_path):
        print(f"Error: The folder '{folder_path}' does not exist.")
        return

    converted_count = 0
    
    for filename in os.listdir(folder_path):
        if filename.lower().endswith(valid_extensions):
            source_path = os.path.join(folder_path, filename)
            
            # Create the new filename with .webp extension
            file_raw_name = os.path.splitext(filename)[0]
            target_path = os.path.join(folder_path, f"{file_raw_name}.webp")
            
            try:
                with Image.open(source_path) as img:
                    # RGB conversion is required for PNGs with transparency if saving to standard JPEG, 
                    # but WebP handles transparency automatically. 
                    img.save(target_path, "WEBP", quality=quality_setting)
                
                print(f"Converted: {filename} -> {file_raw_name}.webp (Quality: {quality_setting})")
                converted_count += 1
                
                if delete_old:
                    os.remove(source_path)
                    
            except Exception as e:
                print(f"Failed to convert {filename}: {e}")
                
    print(f"\nDone! Successfully converted {converted_count} images.")

# Run the function
convert_to_webp(SOURCE_FOLDER, QUALITY, DELETE_ORIGINAL)
```
